El método de Newton es una técnica iterativa basada en el cálculo diferencial que permite encontrar aproximaciones muy precisas de las raíces de una función derivable

A partir de una aproximación inicial $p_0$, el método utiliza la recta tangente a la curva en ese punto para encontrar una nueva aproximación $p_n$ donde la tangente interseca al eje horizontal. La fórmula general de iteración es

$$p_n = p_{n-1} - \frac{f(p_{n-1})}{f'(p_{n-1})}, \quad \text{para } n \ge 1$$

---

## Ejemplo 1

De acuerdo con el problema planteado en el libro, se considera la función:
$$f(x) = \cos(x) - x = 0$$

Para aplicar el algoritmo de Newton, es necesario calcular analíticamente su derivada
$$f'(x) = -\sin(x) - 1$$

```python
# Importo las librerias que necesitamos
import numpy as np
import matplotlib.pyplot as plt

# 1. Definición de la función f(x)
def f(x):
    """Calcula el valor de f(x) = cos(x) - x"""
    return np.cos(x) - x

# 2. Definición de la derivada f'(x)
def df(x):
    """Calcula la derivada f'(x) = -sin(x) - 1"""
    return -np.sin(x) - 1

# 3. Implementación del Método de Newton
def metodo_newton(p0, tol=1e-7, max_iter=100):
    """
    Ejecuta el método de Newton-Raphson para aproximar una raíz.

    Comentarios sobre la sintaxis de Python utilizada:
    - 'def': Palabra reservada para definir funciones personalizadas.
    - 'range(1, max_iter + 1)': Genera una secuencia de números enteros para los ciclos de iteración.
    - 'if / else': Estructuras de control condicional para validar errores (división por cero) y convergencia.
    - 'historial.append(p)': Método de listas para agregar dinámicamente cada aproximación calculada.
    """
    historial = [p0]  # Lista para almacenar el progreso de las aproximaciones

    print(f"{'Iteración':<10} {'p_n':<15} {'f(p_n)':<15}")
    print("-" * 40)

    for n in range(1, max_iter + 1):
        fx = f(p0)
        dfx = df(p0)

        if dfx == 0:
            print("Error: La derivada es cero. El método no puede continuar.")
            break

        # Aplicación de la fórmula de Newton
        p = p0 - fx / dfx
        historial.append(p)

        print(f"{n:<10} {p:<15.7f} {f(p):<15.7e}")

        if abs(p - p0) < tol:
            print("-" * 40)
            print(f"¡Convergencia alcanzada con éxito en la iteración {n}!")
            return p, historial

        # Actualizamos el valor anterior para la siguiente iteración
        p0 = p

    print("Advertencia: Se alcanzó el número máximo de iteraciones permitidas.")
    return p, historial

# Parámetro inicial (aproximación inicial p0 = 0.5)
p0_inicial = 0.5
raiz_aprox, historial_puntos = metodo_newton(p0_inicial)

print(f"\nRaíz aproximada final obtenida: {raiz_aprox:.8f}")

x_vals = np.linspace(0, 1.5, 400)
y_vals = f(x_vals)

plt.figure(figsize=(10, 6))
# Graficamos la función principal en color azul
plt.plot(x_vals, y_vals, label='$f(x) = \\cos(x) - x$', color='blue', linewidth=2)
# Línea horizontal en y = 0 para identificar el cruce por el eje x
plt.axhline(0, color='black', linestyle='--', linewidth=1)

# Graficamos con puntos rojos las aproximaciones generadas en cada iteración de Newton
y_historial = [f(p) for p in historial_puntos]
plt.scatter(historial_puntos, y_historial, color='red', zorder=5, label='Iteraciones de Newton ($p_n$)')

plt.title('Método de Newton - Ejemplo 1 ($\cos(x) - x = 0$)')
plt.xlabel('$x$')
plt.ylabel('$f(x)$')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```

## Solución Final

* *Convergencia cuadrática:* El método de Newton destaca por su velocidad de convergencia. En este ejemplo, el algoritmo alcanza la tolerancia deseada de $10^{-7}$ en apenas *3 o 4 iteraciones*
* *Valor de la raíz:* La solución final obtenida es aproximadamente $x \approx 0.7390851$, lo cual significa que $\cos(0.7390851) \approx 0.7390851$, resolviendo de manera exacta el problema del punto fijo planteado originalmente
* *Visualización:* En la gráfica generada, es posible apreciar cómo los puntos rojos convergen rápidamente hacia la intersección de la curva con el eje horizontal ($f(x) = 0$).
