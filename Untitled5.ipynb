**Problema planteado:** encontrar una raiz aproximando con el metodo de punto fijo a la ecuación:

$$
f(x)=x^3+4x^2-10=0
$$

Se utilizara una aproximación inicial de $x_0=1.5$ y una tolerancia de $10^{-4}$



##2 Fundamento

El método de punto fijo consiste en expresar una ecuación en la forma

$$
x=g(x)
$$

A partir de un valor inicial, se calcula una nueva aproximación mediante la fórmula

$$
x_{n+1}=g(x_n)
$$

El procedimiento se repite hasta que la diferencia entre dos aproximaciones consecutivas sea menor que la tolerancia establecida

Para la ecuación:

$$
x^3+4x^2-10=0
$$

Despejamos $x$:

$$
x^2(x+4)=10
$$

$$
x=\sqrt{\frac{10}{x+4}}
$$

Por tanto, la función iterativa que se utilizará es

$$
\ g(x)=\sqrt{\frac{10}{x+4}}
$$

El error aproximado se calcula mediante

$$
E_n=|x_{n+1}-x_n|
$$

El proceso termina cuando

$$
E_n<10^{-4}
$$

```python
import matplotlib.pyplot as plt

# Funcion de punto fijo
def g(x):
    return (10 / (x + 4))**0.5

# Damos iniciales
x0 = 1.5
tolerancia = 1e-4
max_iteraciones = 100

# guardar los resultados
aproximaciones = [x0]
errores = []

print("METODO DE PUNTO FIJO")
print("Ecuacion: x^3 + 4x^2 - 10 = 0")
print("-" * 58)

# Encabezado de la tabla
print(f"{'n':<6}{'x_n':<16}{'x_n+1':<16}{'Error':<16}")
print("-" * 58)

# Ciclo que realiza las iteraciones
for n in range(1, max_iteraciones + 1):

    # Calcular la nueva aproximacion y el error
    x1 = g(x0)
    error = abs(x1 - x0)

    # Mostrar los resultados con seis decimales
    print(f"{n:<6}{x0:<16.6f}{x1:<16.6f}{error:<16.6f}")

    # Guardar los valores para la grafica
    aproximaciones.append(x1)
    errores.append(error)

    # Comprobamos si se cumple la tolerancia
    if error < tolerancia:
        break

    # Actualizamos el valor para la siguiente iteracion
    x0 = x1

# Mostrar el resultado final
print("\nRESULTADO FINAL")
print(f"Raiz aproximada: {x1:.8f}")
print(f"Error final: {error:.8f}")
print(f"Iteraciones realizadas: {n}")

# Grafica convergencia
plt.figure(figsize=(9, 5))
plt.plot(
    range(len(aproximaciones)),
    aproximaciones,
    color="royalblue",
    marker="o",
    markersize=5,
    linewidth=2,
    label="Aproximaciones"
)
plt.axhline(
    y=x1,
    color="crimson",
    linestyle="--",
    label=f"Raiz aproximada = {x1:.6f}"
)

plt.title("Convergencia del metodo de punto fijo")
plt.xlabel("Numero de iteracion")
plt.ylabel("Valor de x")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()

# Grafica del error
plt.figure(figsize=(9, 5))
plt.semilogy(
    range(1, len(errores) + 1),
    errores,
    color="darkorange",
    marker="s",
    linewidth=2
)
plt.axhline(
    y=tolerancia,
    color="green",
    linestyle="--",
    label="Tolerancia = 0.0001"
)

plt.title("Disminucion del error por iteracion")
plt.xlabel("Numero de iteracion")
plt.ylabel("Error absoluto (escala logaritmica)")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

##Solución final

La raíz aproximada de la ecuación

$$
x^3+4x^2-10=0
$$

que obtuvimos con el método de punto fijo es:

$$
{x\approx 1.365230}
$$


##Interpretación de los resultados

En cada iteración se utiliza el valor anterior para calcular una nueva aproximación. Al principio, la diferencia entre ambos valores es relativamente grande, pero después es menor.

La primera gráfica permite observar cómo las aproximaciones se acercan a un valor estable. La segunda muestra la reducción del error y permite compararlo con la tolerancia de $10^{-4}$.

El programa se detiene cuando el error entre dos aproximaciones consecutivas es menor que la tolerancia que definimos.
